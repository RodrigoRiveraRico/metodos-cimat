### Solución:

En general, el método de Gauss-Seidel es:
$$x_i^{(t+1)} = \frac{1}{a_{i,i}}\left(b_i-\sum_{j=1}^{i-1}a_{i,j}x_j^{(t-1)}-\sum_{j=i+1}^{n}a_{i,j}x_j^{(t)}\right)$$

Para una matriz $A$ tridiagonal:

Para $i=1$: Notamos que $a_{1,j}=0$ para $j>2$.
$$x_1^{(t+1)}=\frac{1}{a_{1,1}}\left(b_1-a_{1,2}x_2^{(t)}\right)$$

Para $1<i<n$: Notamos que $a_{i,j}\neq0$ para $i-1\leq j\leq i+1$.
$$x_i^{(t+1)}=\frac{1}{a_{i,i}}\left(b_i-a_{i,i-1}x_{i-1}^{(t+1)}-a_{i,i+1}x_{i+1}^{(t)}\right)$$

Para $i=n$: Notamos que $a_{n,j}\neq0$ para $j\geq n-1$.
$$x_n^{(t+1)}=\frac{1}{a_{n,n}}\left(b_n-a_{n,n-1}x_{n-1}^{(t+1)}\right)$$

Ahora hagamos obtengamos las componente del vector $v=Ax$:

Para $i=1$:
$$v_1=a_{1,1}x_1+a_{1,2}x_2$$

Para $1<i<n$:
$$v_i=a_{i,i-1}x_{i-1}+a_{i,i}x_i+a_{i,i+1}x_{i+1}$$

Para $i=n$:
$$v_{n}=a_{n,n-1}x_{n-1}+a_{n,n}x_n$$
